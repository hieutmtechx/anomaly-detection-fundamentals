```python
import pandas as pd

# Đọc file HDF5
df_test = pd.read_hdf("../../Data/phase2_ground_truth.hdf")

# Xem cấu trúc dữ liệu
print(df_test.head())
print(df_test.columns)
print(df_test.dtypes)
print(df_test.shape)
```

```python
df_test["KPI ID"] = df_test["KPI ID"].astype(str)
print(df_test["label"].value_counts())  # xem tỷ lệ bất thường/bình thường
```

```python
df_data = pd.read_csv("../../Data/test.csv")  # file dữ liệu test (tên hơi gây nhầm lẫn)
df_data["KPI ID"] = df_data["KPI ID"].astype(str)

merged = df_data.merge(df_test, on=["timestamp", "KPI ID"], suffixes=("", "_truth"))
```
